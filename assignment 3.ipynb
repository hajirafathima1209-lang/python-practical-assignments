```python
1.Sum of Even and Odd Numbers Get a number n from the user 
and calculate the sum of all even numbers and the sum of all 
odd numbers between 1 and n.
```

```python
n = int(input("Enter a number: "))

even_sum = 0
odd_sum = 0

for i in range(1, n + 1):
    if i % 2 == 0:
        even_sum += i
    else:
        odd_sum += i

print("Sum of even numbers:", even_sum)
print("Sum of odd numbers:", odd_sum)
```

```python
2.Count Digits in a Number Get an integer from the user and count the number
of digits using a single for loop. Do not convert the number into a string. Example: 58392 → 5 digits
```

```python
n = abs(int(input("Enter a number: ")))
count = 0

for i in range(100):
    if i == 0 and n == 0:
        count = 1
        break

    if n == 0:
        break

    count += 1
    n //= 10

print("Number of digits:", count)
```

```python
3. Find the Second-Largest Number Given a list of distinct numbers, 
find the second-largest number without using sort() or max().
```

```python
numbers = [10, 25, 45, 30, 50]

largest = numbers[0]
second = numbers[1]

if second > largest:
    largest, second = second, largest

for i in range(2, len(numbers)):
    if numbers[i] > largest:
        second = largest
        largest = numbers[i]
    elif numbers[i] > second:
        second = numbers[i]

print("Second largest number:", second)
```

```python
4. Count Vowels and Consonants Get a string from the user and count the vowels 
and consonants separately. Ignore spaces, digits, and special characters.
```

```python
s = input("Enter a string: ")

vowels = 0
consonants = 0

for ch in s:
    if ch.isalpha():
        if ch.lower() in "aeiou":
            vowels += 1
        else:
            consonants += 1

print("Vowels:", vowels)
print("Consonants:", consonants)
```

```python
5.Calculate Factorial Get a positive integer from the user and 
calculate its factorial using a single for loop. Example: 5 → 120
```

```python
n = int(input("Enter a positive number: "))

fact = 1

for i in range(1, n + 1):
    fact = fact * i

print("Factorial:", fact)
```

```python
6.Find Numbers Divisible by 7 but Not by 5 Print all numbers between 1 and
200 that are divisible by 7 but not divisible by 5. Also, count the matching numbers.
```

```python
count = 0

for i in range(1, 201):
    if i % 7 == 0 and i % 5 != 0:
        print(i, end=" ")
        count += 1

print("\nTotal count:", count)
```

```python
7.Calculate Electricity Bill Get the units consumed by each of five customers using a list. 
Calculate each customer's bill using these rates:

* First 100 units: ₹2 per unit
* Next 100 units: ₹3 per unit
* Above 200 units: ₹5 per unit
```

```python
units = []

for i in range(5):
    u = int(input("Enter units consumed: "))
    units.append(u)

for i in range(5):
    u = units[i]

    if u <= 100:
        bill = u * 2
    elif u <= 200:
        bill = 100 * 2 + (u - 100) * 3
    else:
        bill = 100 * 2 + 100 * 3 + (u - 200) * 5

    print("Customer", i + 1, "Bill: ₹", bill)
```

```python
8.Reverse a Number Get an integer from the user and 
reverse its digits using a single for loop. Do not convert it into a string. Example: 12345 → 54321
```

```python
n = int(input("Enter a number: "))

num = abs(n)
reverse = 0

for i in range(100):
    if num == 0:
        break

    digit = num % 10
    reverse = reverse * 10 + digit
    num //= 10

if n < 0:
    reverse = -reverse

print("Reversed number:", reverse)
```

```python
9.Analyze Student Marks Given a list of marks, count the students who passed (50 or above), 
failed (below 50), and scored distinction (75 or above).
```

```python
marks = [85, 45, 90, 35, 75, 60, 40]

passed = 0
failed = 0
distinction = 0

for mark in marks:
    if mark >= 50:
        passed += 1
    else:
        failed += 1

    if mark >= 75:
        distinction += 1

print("Passed students:", passed)
print("Failed students:", failed)
print("Distinction students:", distinction)
```

```python
10.Find Perfect Numbers Print all perfect numbers between 1 and 1000 using a single for loop. 
A perfect number equals the sum of its proper divisors. Example: 6 is perfect because 1 + 2 + 3 = 6.
```

```python
for num in range(1, 1001):
    total = 0

    for i in range(1, num):
        if num % i == 0:
            total += i

    if total == num:
        print(num)
```
