```python
1. Leap Year Checker
Accept a year. First check whether it is divisible by 4. If yes, check whether it is divisible by 100.
If it is, check whether it is divisible by 400. Display whether the year is a leap year.
```

```python
year = int(input("Enter a year: "))

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print("Leap year")
        else:
            print("Not a leap year")
    else:
        print("Leap year")
else:
    print("Not a leap year")
```

```python
2. Triangle Classification
Accept three side lengths. First check whether they can form a valid triangle. 
If valid, check whether all three sides are equal, exactly two sides are equal, or all sides are different.
```

```python
a = int(input("Enter first side: "))
b = int(input("Enter second side: "))
c = int(input("Enter third side: "))

if a + b > c and a + c > b and b + c > a:
    if a == b and b == c:
        print("Equilateral triangle")
    elif a == b or b == c or a == c:
        print("Isosceles triangle")
    else:
        print("Scalene triangle")
else:
    print("Invalid triangle")
```

```python
3. Employee Bonus Calculator
Accept years of experience and performance rating. If experience is at least 5 years, check the rating. 
Give a 20% bonus for an excellent rating and 10% for a good rating. 
If experience is below 5 years, give a 5% bonus only for an excellent rating.
```

```python
experience = int(input("Enter years of experience: "))
rating = input("Enter performance rating (excellent/good/average): ")

if experience >= 5:
    if rating == "excellent":
        print("Bonus: 20%")
    elif rating == "good":
        print("Bonus: 10%")
    else:
        print("Bonus: 0%")
else:
    if rating == "excellent":
        print("Bonus: 5%")
    else:
        print("Bonus: 0%")
        
```

```python
4. Insurance Premium Category
Accept age, smoking status, and annual income. If the applicant is under 30,
check smoking status to determine the risk category. 
If age is 30 or above, check income and smoking status to classify the applicant as low, medium, or high risk. 
Define your own clear rules for each category.
```

```python
age = int(input("Enter age: "))
smoking = input("Are you a smoker? (yes/no): ")
income = float(input("Enter annual income: "))

if age < 30:
    if smoking == "yes":
        print("Medium risk")
    else:
        print("Low risk")
else:
    if income >= 500000:
        if smoking == "no":
            print("Low risk")
        else:
            print("Medium risk")
    else:
        if smoking == "yes":
            print("High risk")
        else:
            print("Medium risk")
```

```python
15. Hospital Consultation Priority
Accept a patients temperature and oxygen saturation. 
If oxygen saturation is below 90%, classify as high priority. 
Otherwise, check whether the temperature is at least 39°C. If so, classify as urgent; otherwise, classify as normal priority.
(Simplified programming exercise, not a medical triage tool.)
```

```python
temperature = float(input("Enter temperature in Celsius: "))
oxygen = float(input("Enter oxygen saturation: "))

if oxygen < 90:
    print("High priority")
else:
    if temperature >= 39:
        print("Urgent")
    else:
        print("Normal priority")
```

```python
16. Income Tax Slab Calculator
Accept annual taxable income. If income is up to ₹3 lakh, tax is 0%.
Otherwise, check whether it is up to ₹7 lakh and apply a 5% rate to income above ₹3 lakh. 
For income above ₹7 lakh, apply progressive slabs of 0%, 5%, and 10% to the respective portions. Calculate the total tax.
```

```python
income = float(input("Enter annual taxable income: "))

if income <= 300000:
    tax = 0
elif income <= 700000:
    tax = (income - 300000) * 0.05
else:
    tax = (400000 * 0.05) + ((income - 700000) * 0.10)

print("Total tax:", tax)
```

```python
17. College Admission
Accept marks in Mathematics, Physics, and Chemistry. If Mathematics is at least 60, check Physics.
If Physics is at least 50, check Chemistry. If Chemistry is at least 50 and the total is at least 180, 
display "Eligible"; otherwise, display "Not Eligible".
```

```python
maths = int(input("Enter Mathematics marks: "))
physics = int(input("Enter Physics marks: "))
chemistry = int(input("Enter Chemistry marks: "))

total = maths + physics + chemistry

if maths >= 60:
    if physics >= 50:
        if chemistry >= 50 and total >= 180:
            print("Eligible")
        else:
            print("Not Eligible")
    else:
        print("Not Eligible")
else:
    print("Not Eligible")
```
